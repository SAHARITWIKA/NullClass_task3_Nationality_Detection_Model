```python
import os
import numpy as np
import pandas as pd
import cv2
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import *
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
```

```python

# 1. Load preprocessed age+image data
import os
import cv2
import numpy as np

base_path = "../data/imdb_crop"  
images = []
ages = []

for folder in os.listdir(base_path):
    folder_path = os.path.join(base_path, folder)
    if not os.path.isdir(folder_path):
        continue

    for file in os.listdir(folder_path):
        if file.endswith(".jpg"):
            try:
                birth_year = int(file.split("_")[2].split("-")[0])
                photo_year = int(file.split("_")[3].split(".")[0])
                age = photo_year - birth_year

                if age < 0 or age > 100:
                    continue

                img_path = os.path.join(folder_path, file)
                img = cv2.imread(img_path)
                if img is None:
                    continue

                img = cv2.resize(img, (128, 128))
                images.append(img)
                ages.append(age)

            except Exception as e:
                print(f"Skipping {file}: {e}")

X = np.array(images) / 255.0
y = np.array(ages)

print(f"✅ Total images loaded: {len(X)}")
print(f"✅ Total labels loaded: {len(y)}")
```

```python
print(f"Total images loaded: {len(X)}")
print(f"Total labels loaded: {len(y)}")
```

```python
# 2. Train/Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

```

```python
# 3. Regression Model
model = Sequential([
    Conv2D(32, (3,3), activation='relu', input_shape=(224,224,3)),
    MaxPooling2D(),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(),
    Flatten(),
    Dense(128, activation='relu'),
    Dense(1)
])
model.compile(optimizer='adam', loss='mean_squared_error')
model.fit(X_train, y_train, validation_data=(X_test, y_test), epochs=10)
```

```python
# 4. Evaluation
y_pred = model.predict(X_test).flatten()
mae = mean_absolute_error(y_test, y_pred)
print(f"Mean Absolute Error: {mae}")

with open("../evaluation_metrics/age_model_metrics.txt", "w") as f:
    f.write(f"MAE: {mae}")
```

```python
# 5. Save Model
model.save("../models/age_model.h5")
```
