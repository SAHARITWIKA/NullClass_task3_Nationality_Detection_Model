```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# ➕ Ensure utils folder is in the Python path
import sys
sys.path.append('./utils')  # Adjust if needed

from preprocess import load_and_resize_image, extract_dominant_color, predict_color_name

# 🔍 Load image
image_path = "../data/sample_dress.jpg"
try:
    image = Image.open(image_path).convert("RGB").resize((224, 224))
    image_np = np.array(image)

    # 🎯 Get dominant color
    dominant_rgb = extract_dominant_color(image_np)
    color_name = predict_color_name(dominant_rgb)

    # 📊 Show image with prediction
    plt.imshow(image_np)
    plt.title(f"Predicted Dress Color: {color_name} ({dominant_rgb})")
    plt.axis('off')
    plt.show()

except FileNotFoundError:
    print(f"❌ Image not found at path: {image_path}")
```
